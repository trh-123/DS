```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report


df = pd.read_csv("HR_comma_sep.csv")
df.head()

X = df[['satisfaction_level', 'average_montly_hours']]
y = df['left']

# Mengubah 'Sex' menjadi numerik (0: female, 1: male)
#X['sex'] = X['sex'].map({'female': 0, 'male': 1})

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)


knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)


new_employee = pd.DataFrame([[0.5, 100]], columns=['satisfaction_level','average_montly_hours'])


new_data_scaled = scaler.transform(new_employee)

prediction = knn.predict(new_data_scaled)[0]

print("Prediksi:", "left" if prediction==1 else "NOT left")


acc = accuracy_score(y_test, y_pred)
print("\nAkurasi Model:", acc)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))



```
