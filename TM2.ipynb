```python
#1: Import libraries
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.mixture import GaussianMixture

#2: Load data & exploratory analysis 
df=pd.read_csv('titanic.csv')
df
```

```python
#2: Load data & exploratory analysis 

df.describe()
```

```python
#3: Deteksi masalah data
df.isnull().sum()
```

```python
#4: Imputasi data
# Imputasi hanya kolom numerik
num_cols = df.select_dtypes(include='number').columns

imputer = SimpleImputer(strategy='mean')
df_imputed = df.copy()
df_imputed[num_cols] = imputer.fit_transform(df[num_cols])

# Fit GMM pada kolom fare (yang sudah diimputasi)
gmm = GaussianMixture(n_components=2, max_iter=100, random_state=42)
gmm.fit(df_imputed[['fare']])

# Menampilkan DataFrame hasil imputasi
df_imputed.head()
```
